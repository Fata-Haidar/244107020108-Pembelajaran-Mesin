**Memisahkan antara variabel yang dapat digunakan dan variabel yang tidak dapat digunakan.**

```python
import pandas as pd

# Load dataset
df = pd.read_csv("wbc.csv")

# Melihat informasi dataset
print(df.head())
print(df.info())


# Variabel tidak digunakan
unused_variables = []

# biasanya kolom id dan Unnamed tidak diperlukan
if "id" in df.columns:
    unused_variables.append("id")

if "Unnamed: 32" in df.columns:
    unused_variables.append("Unnamed: 32")


# Hapus variabel tidak digunakan
df_clean = df.drop(
    columns=unused_variables,
    errors="ignore"
)


print("Variabel tidak digunakan:")
print(unused_variables)


print("\nVariabel yang digunakan:")
print(df_clean.columns)
```

**Melakukan proses encoding pada kolom "diagnosis".**

```python
from sklearn.preprocessing import LabelEncoder


encoder = LabelEncoder()

df_clean["diagnosis"] = encoder.fit_transform(
    df_clean["diagnosis"]
)


print("Hasil encoding diagnosis:")
print(df_clean["diagnosis"].head())


print("\nMapping:")
for label, value in zip(
    encoder.classes_,
    encoder.transform(encoder.classes_)
):
    print(label, "=", value)
```

**Melakukan proses standardisasi pada semua kolom yang memiliki nilai numerik.**

```python
from sklearn.preprocessing import StandardScaler


# Pisahkan fitur dan target

X = df_clean.drop(
    "diagnosis",
    axis=1
)

y = df_clean["diagnosis"]


# Standardisasi
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)


# ubah kembali menjadi dataframe
X_scaled = pd.DataFrame(
    X_scaled,
    columns=X.columns
)


print(X_scaled.head())
```

**Melakukan proses seleksi fitur. Anda dapat menggunakan SelectKBest.**

```python
from sklearn.feature_selection import SelectKBest, f_classif


# mencoba semua jumlah fitur

feature_results = []


for k in range(1, X.shape[1]+1):

    selector = SelectKBest(
        score_func=f_classif,
        k=k
    )


    X_selected = selector.fit_transform(
        X_scaled,
        y
    )


    feature_results.append(
        {
            "jumlah_fitur": k,
            "fitur": X_selected.shape[1]
        }
    )


hasil_fitur = pd.DataFrame(
    feature_results
)


print(hasil_fitur)
```

**Melakukan proses pengujian dengan model Logistic Regression seperti pada praktikum 1 dengan menggunakan pipeline.**

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.feature_selection import SelectKBest, f_classif
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report


# Membuat pipeline

pipeline_model = Pipeline([
    ("scaler", StandardScaler()),

    ("selector", SelectKBest(
        score_func=f_classif,
        k=5
    )),

    ("clf", LogisticRegression(
        max_iter=1000
    ))
])


# Training pipeline

pipeline_model.fit(
    X_train,
    y_train
)


# Prediksi

y_pred = pipeline_model.predict(
    X_test
)


# Evaluasi model

print("=== Pipeline SelectKBest + Logistic Regression ===")

print(
    "Accuracy:",
    accuracy_score(
        y_test,
        y_pred
    )
)


print("\nClassification Report:")
print(
    classification_report(
        y_test,
        y_pred
    )
)
```

**Jumlah fitur terbaik yang dapat digunakan dan Apa saja fitur tersebut**

```python
hasil = []


# Uji semua jumlah fitur

for k in range(1, X.shape[1] + 1):

    pipeline = Pipeline([

        ("scaler", StandardScaler()),

        ("selector", SelectKBest(
            score_func=f_classif,
            k=k
        )),

        ("clf", LogisticRegression(
            max_iter=1000
        ))
    ])


    pipeline.fit(
        X_train,
        y_train
    )


    prediksi = pipeline.predict(
        X_test
    )


    accuracy = accuracy_score(
        y_test,
        prediksi
    )


    hasil.append(
        [
            k,
            accuracy
        ]
    )



# Tabel hasil pengujian

hasil_df = pd.DataFrame(
    hasil,
    columns=[
        "Jumlah Fitur",
        "Accuracy"
    ]
)


print("Hasil pengujian jumlah fitur:")
print(hasil_df)



# Mengambil jumlah fitur dengan accuracy tertinggi

best_k = hasil_df.loc[
    hasil_df["Accuracy"].idxmax(),
    "Jumlah Fitur"
]


best_accuracy = hasil_df.loc[
    hasil_df["Accuracy"].idxmax(),
    "Accuracy"
]


print("\nJumlah fitur terbaik:")
print(best_k)


print("\nAccuracy terbaik:")
print(best_accuracy)



# ==========================================
# Menampilkan nama fitur terbaik
# ==========================================

best_selector = SelectKBest(
    score_func=f_classif,
    k=int(best_k)
)


best_selector.fit(
    X,
    y
)


selected_features = X.columns[
    best_selector.get_support()
]


print("\nFitur yang digunakan:")
for feature in selected_features:
    print("-", feature)
```
