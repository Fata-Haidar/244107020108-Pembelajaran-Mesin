```python
import pandas as pd
from sklearn.preprocessing import StandardScaler, LabelEncoder


# 1. Load dataset & hapus kolom yang tidak dapat digunakan
dtugas = 'wbc.csv'
df = pd.read_csv(dtugas)
df_clean = df.drop(columns=['id', 'Unnamed: 32'])

# 2. Encoding target 'diagnosis'
le = LabelEncoder()
df_clean['diagnosis'] = le.fit_transform(df_clean['diagnosis'])

# 3. Standardisasi fitur numerik
features = df_clean.drop(columns=['diagnosis']).columns
scaler = StandardScaler()
```

```python
df["diagnosis"] = df["diagnosis"].map({
    "B": 0,
    "M": 1
})

print("\nHasil encoding diagnosis:")
print(df["diagnosis"].value_counts())
```

```python
kolom_numerik = df.select_dtypes(include=["int64", "float64"]).columns
kolom_numerik = kolom_numerik.drop("diagnosis")

# StandardScaler
scaler = StandardScaler()

df[kolom_numerik] = scaler.fit_transform(df[kolom_numerik])

print("\nData setelah standardisasi:")
print(df.head())
```

```python
df.to_csv("wbc_preprocessed.csv", index=False)
```
